## Step 5: Problem Definition

### The Transverse-field Ising Model

In the Transverse-field Ising model, each site can have a spin that can be in one of two states: either +1 (spin-up) or -1 (spin-down).

The Hamiltonian is given by:

$H = -J \sum_{<i,j>}{S_i S_j} + b\sum_i S_i$

Where:
- $J$ is the interaction strength between neighboring spins
- $<i,j>$ are the nearest neighbors
- $S_i$ is the spin at site $i$
- $b$ is the relative strength of the external field

For a simple 2-spins system:

$H = -J S_1 S_2 + b(S_1 + S_2)$
